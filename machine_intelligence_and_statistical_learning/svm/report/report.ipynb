### Θεωρία SVM
Η μηχανή SVM είναι μια αρκετά δημοφιλής μηχανή επιβλεπόμενης μάθηση, που παρόλο την άνοδο και την επικράτηση των νευρωνικών δικτύων, χρησιμοποιείται αρκετά σήμερα. Η μηχανή είναι ιδιαίτερα χρήσιμη όταν το πρόβλημα έχει λίγα δεδομένα διαχωρίζοντας τα δεδομένα χρησιμοποιώντας ένα υπερεπίπεδο. Αν από τα χαρακτηριστικά των δεδομέων που βρίσκονται στο πρόβλημα δεν επαρκούν για την επίλυση του προβλήματος, τότε μπορούν να χρησιμοποιηθούν τεχνικές ώστε να βρεθεί ένα επίπεδο σε μία μεγαλύτερη διάσταση χρησιμοποιώντας την τεχνική "kernel trick". Το τελικό επίπεδο που χρησιμοποιείται δεν είναι τίποτα άλλο παρά την προβολή του υπερεπιπέδου που βρέθηκε στην μεγαλύτερη διάσταση στην μικρότερη.

#### Μαθηματική μοντελοποίηση
Έστω μια συλλογή δεδομένων n-διαστάσεων και δύο κλάσσεων η οποία θα θεωρηθεί ότι είναι τέλεια διαχωρίσιμη. Έστω επίπεδο με κάθετο διάνυσμα $w$ και σταθερό όρο $b$ τότε ισχύει ότι:

\begin{align}
w^T\cdot x_i + b \leq -1 \tag{1}
\end{align}

\begin{align}
w^T\cdot x_j + b \geq +1 \tag{2}
\end{align}


Σε αυτό το σετ εξισώσεων οι άγνωστοι είναι το $w$ και το $b$, όπου το $w$ είναι ένα διάνυσμα n-διαστάσεων ενώ το b ένας πραγματικός αριθμός. Επιδιώκεται οπότε να βρεθεί ένα w που χωρίζει τον υπερχώρο τέλεια στα δύο. Έστω ότι επιλέγεται ένα σημείο στο θετικό υπερεπίπεδο και ένα στο αρνητικό. Τότε ισχύει ότι
\begin{align}
w^T\cdot x^{-} + b = -1 \tag{3} 
\end{align}

\begin{align}
w^T\cdot x^{+} + b = +1 \tag{4}
\end{align}

Θεωρείται απόσταση $M$ όπου είναι η απόσταση μεταξύ των δύο υπερεπιπέδων.
\begin{align}
|\lambda w| = M \tag{5}
\end{align}

\begin{align}
x^+ &= x^- + \lambda *w \tag{6}
\end{align}

Από τις σχέσεις αυτές βγαίνει ότι
\begin{align}
Μ &= \frac{2}{\sqrt{w^T\cdot w}} \tag{7}
\end{align}

Στο πρόβλημα πρέπει το $M$ να μεγιστοποιηθεί ή η ποσότητα $w^T\cdot w$ να ελαχιστοποιηθεί ικανοποιώντας ότι τα σημεία θα πρέπει να βρίσκονται από την μεριά του αντίστοιχου υπερεπιπέδου

\begin{align}
y_i( w^t\cdot x_i + b ) \geq 1 \tag{8}
\end{align}

Σε αυτό το πρόβλημα τα δεδομένα είναι τέλεια διαχωρίσιμα, αλλά οι περισσότερες εφαρμογές αν όχι όλες τα παραδείγματα δεν είναι τέλεια διαχωρίσιμα. Οπότε, εισάγεται όρος $C$ και τα λάθη ως $\xi _i$ για το κάθε δείγμα. Οπότε αξιοποιώντας και τις ανοχές αυτές, ισχύει ότι:
\begin{equation}
\min_{w, b, \xi _i} \frac{w_T\cdot w}{2} + C \sum _{i=0} ^n \xi _{i} \tag{9} \\
\end{equation}

$$
\text{subject to }
$$

\begin{equation}
y_i(w^T\cdot x_i + b ) \geq 1 - \xi _i \tag{10}\\
\end{equation}

\begin{equation}
\xi _i \geq 0 \tag{11}
\end{equation}

Όπου αυτό αποτελεί ένα πρόβλημα γραμμικού προγραμματισμού. Όπου το πρόβλημα αυτό μετασχηματίζεται στο δυϊκό πρόβλημα Lagrange.

\begin{equation}
\min _{\bold{\alpha}} \frac{\alpha ^T Q \alpha}{2} - e^T\cdot \alpha \tag{12}
\end{equation}

$$
\text{subject to }
$$

\begin{equation}
y^T\cdot \alpha = 0 \tag{13}
\end{equation}

\begin{equation}
0 \leq a_i \leq C, i = 1, ...,n \tag{14}
\end{equation}

Όπου $e$ διάνυσμα με μέγεθος $n$ με στοιχεία το 1, ενώ $Q = y_i y_j (x_i^T \cdot x_j)$. Αν χρησιμοποιηθεί το Kernel trick τότε 
\begin{equation}
Q = y_i y_j K( x_i, x_j ) \tag{15}
\end{equation}

\begin{equation}
K( x_i, x_j ) = \phi( x_i )^T \phi( x_j ) \tag{16}
\end{equation}

Όπου $\phi$ είναι μία συνάρτηση πυρήνα και αποτελεί βάση του Hilbert space που χρησιμοποιείται. Από το δυϊκό πρόβλημα επίσης βγαίνει ότι

\begin{equation}
w = \sum_{i = 0}{n} y_i, \alpha_i \phi( x_i ) \tag{17}
\end{equation}

Στις περισσότερες τιμές το $\alpha$ μηδενίζεται ενώ υπάρχουν τιμές που το $\alpha$ έχει τιμή. Στις τιμές όπου το \alpha έχει τιμή το επίπεδο βρίσκεται πάνω στo συγκεκριμένο διάνυσμα $x$.

\begin{equation}
w^T\cdot \phi( x ) + b = y_i \tag{18}
\end{equation}


Μπορεί να χρησιμοποιηθεί μία εξίσωση για να βρεθεί το $b$, για λόγους αριθμητικής ευστάθειας παίρνουμε τον μέσο όρο από όλες τις εξισώσεις που ικανοποιούν την (18).

Το αποτέλεσμα τελικά δίνεται από

\begin{equation}
y_{pred} = \text{sgn} ( w_T \phi(x) + b ) = \text{sgn}( \sum_{i = 0}^{n}( y_i \alpha_i K( x_i, x ) + b ) ) \tag{19}
\end{equation}

$$
\text{ή στον δυικό χώρο }
$$

\begin{equation}
b = \frac{1}{N_{margin}} \sum_{s \in S_{margin}} \left( y_s - \sum_{i \in SV} \alpha_i y_i K(x_i, x_s) \right) \tag{20}
\end{equation}

Με αυτό τον τρόπο δεν χρειάζεται η μετάβαση στον βασικό χώρο. Σε γενικές γραμμές είναι προτιμότερο να υπολογίζεται η πρόβλεψη του ταξινομητή στον βασικό χώρο όταν το πρόβλημα έχει πάρα πολλά δεδομένα αλλά λίγες διαστάσεις. 


Μερικές από τις συναρτήσεις πυρήνα είναι
\begin{equation}
\text{linear}\quad K( x, x_i )= x_i\cdot x \tag{21}
\end{equation}

\begin{equation}
\text{polynomial}\quad K( x, x_i ) = ( x_i^T x + \tau )^d \tag{22}
\end{equation}

\begin{equation}
\text{RBF}\quad K(x_i, x_j) = \exp\left(-\gamma \|x_i - x_j\|^2\right) \tag{23} 
\end{equation}


## Ταξινομητής για CIFAR-10
Η βάση δεδομένων περιέχει διαφορετικές εικόνες από 10 διαφορετικές κατηγορίες. Για κάθε κατηγορία περιέχονται 5000 εικόνες ενώ έχει και ένα σύνολο δεδομένων για επικύρωση τα οποία αριθμούν στις 10000. Η κάθε εικόνα είναι μεγέθους 32 επί 32 και περιέχει τρία χρώματα.

### Διαδικασία προεπεξεργασίας
#### Μετατροπή σε ασπρόμαυρο
Η πρώτη διαδικασία είναι η μετατροπή τους σε ασπρόμαυρες. Αυτό υλοποίηται χρησιμοποιώντας τον μετασχηματισμό
\begin{equation}
\text{Grayscale} = 0.299R + 0.587G + 0.114B\tag{24}
\end{equation}

Αυτό θα βοηθήσει στην μείωση των διαστάσεων από 96 σε 32 επιτρέποντας τον ταξινομητή να εκπαιδευτεί πιο γρήγορα.
#### HOG
Για την προεπεξεργασία χρησιμοποιήθηκε ο HOG(Histogram of Oriented Gradients) feature detection. Τα βήματα το HOG είναι τα εξής
1. Για κάθε pixel της εικόνας υπολογίζεται η παράγωγος ως προς την x και y διεύθυνση
1. Για κάθε εικονοστοιχείο υπολογίζεται το μέτρο της κλίσης και η κλίση σε μοίρες.
1. Χωρίζεται η εικόνα σε μη επικαλυπτούμενα κελιά. Για κάθε κελί κατασκευάζεται ιστόγραμμα με 9 στοιχεία. Σε κάθε κάδο (bin) του ιστογράμματος προστίθεται το μέτρο του διανύσματος που αντιστοιχεί το συγκεκριμένο εικονοστοιχείο. Με αυτόν τον τρόπο απεικονίζεται η εικόνα από 9 αριθμούς.
1. Ακολουθεί μία εξισορρόπηση φωτινότητας άνα μπλοκ.
1. Τελικά η έξοδος του αλγορίθμου είναι οι τιμές των κελιών.

### Υλοποίηση ταξινομητή


#### One vs One
Στην αρχιτεκτονική one vs one εκπαιδεύονται δυαδικοί ταξινομητές. Στην συγκεκριμένη περίπτωση αν n είναι οι ταξινομητές τότε εκπαιδεύονται σύνολο $\frac{n\cdot ( n - 1 )}{2}$. Οι ταξινομητές εκπαιδεύονται μόνο στις κλάσσεις τους. Η τελική απόφαση γίνεται μέσω διαδικασίας ψηφοφορίας όπου και κάθε μηχανή ψηφίζει για το περιεχόμενο της εικόνας. Σαν εναλλακτικό του hard voting, υπάρχει το soft voting. Στην δεύτερη περίπτωση η μηχανή ψηφίζει με βάση την εκτιμώμενη πιθανότητα ότι το αντικείμενο ανήκει στη συγκεκριμένη κλάση.

#### One vs All
Σε αυτήν την περίπτωση ο ταξινόμητης εκπαιδεύεται σε κάθε κατηγορία εναντίον όλη της υπόλοιπης. Η τελική επιλογή της προβλεπόμενης κλάσσης μπορεί να γίνει με την πιθανότητα πρόβλεψης από την κάθε μηχανή.

Σε αυτήν τη περίπτωση επιλέχθηκε η στρατιγική one vs one, όπου και υλοποιήθηκαν 45 διαφορετικοί ταξινομητές. Επίσης, για πειραματισμό χρησιμοποιήθηκε και η τεχνική stacking, όπου αντί για soft voting χρησιμοποιήθηκε ένα μοντέλο γραμμικής παλινδρόμησης με κατωφλίωση για την τελική απόφαση του ταξινομητή χωρίς όμως να φαίνεται να προσδίδει σημαντικά στο αποτέλεσμα.

#### Εύρεση υπερπαραμέτρων
Για την εύρεση των υπερπαραμέτρων χρησιμοποιήθηκε η τεχνική GridSearch η οποία υλοποιήται μέσω της συνάρτησης της scikit learn GridSearchCV. Για την εύρεση υπερπαραμέτρων το πλέγμα είναι για την παράμετρο από το 0.06 μέχρι το 0.6 με βήμα 0.02 και η παράμετρος gamma με 0.01 μέχρι 1. με βήμα .2 το οποίο μας cross validation = 5 δίνει συνολικά ότι πρέπει να γίνουν $135*5 = 675$ εκπαιδεύσεις.

Η υλοποίηση του SVM classifier έχει καλύτερη επίδοση και στις τρεις τεχνικές που χρησιμοποιήθηκαν, συκγριτικά όμως χρειάστηκε πολύ μεγαλύτερος χρόνος για να εκπαιδευτεί και πραγματοποιήθηκε μεγαλύτερη μελέτη για την παραμετροποίηση του ταξινομητή.

## Βελτιώσεις
1. Αξιοποίηση όλων των υπόλοιπον χρωμάτων. Οι εικόνες που εισάγονται είναι ασπρόμαυρες, αυτό έχει ως αποτέλεμα να χάνεται πληροφορία η οποία πιθανώς να ήταν χρήσιμη για τον ταξινομητή.
1. Μελέτη του ταξινομητή one vs all χρησιμοποιώντας την υλοποίηση της cuml.
1. Αλλαγή του Linear regressor με εναν svr. Πιθανώς μπορεί να μπει σε αυτόν τον regressor και μία είσοδο και για την εικόνα.
